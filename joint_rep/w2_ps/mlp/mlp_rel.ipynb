```python
# Device (MPS on Apple Silicon if available)
DEVICE = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
# print(DEVICE)
DEVICE = torch.device('cpu')

# Repro
def set_seed(seed=42):
    np.random.seed(seed)
    torch.manual_seed(seed)

class MLP(nn.Module):
    def __init__(self, in_dim, h1=128, h2=64):
        super().__init__()
        half = in_dim // 2

        # Learnable scaling for both halves
        self.scale_first = nn.Parameter(torch.ones(half))
        # self.scale_second = nn.Parameter(torch.ones(half))

        # Main MLP network
        self.net = nn.Sequential(
            nn.Linear(in_dim, h1),
            nn.ReLU(),
            nn.Linear(h1, h2),
            nn.ReLU(),
            nn.Linear(h2, 1),
        )

    def forward(self, x):
        half = x.shape[1] // 2

        # Split and scale both halves
        first_half = x[:, :half] * self.scale_first
        second_half = x[:, half:] # * self.scale_second

        # Concatenate back
        x = torch.cat([first_half, second_half], dim=1)

        logits = self.net(x)
        return logits.squeeze()

@torch.no_grad()
def _evaluate(model, loader, criterion):
    model.eval()
    all_logits = []
    all_targets = []
    running_loss = 0.0
    n = 0
    for xb, yb in loader:
        xb = xb.to(DEVICE)
        yb = yb.to(DEVICE)
        logits = model(xb)
        loss = criterion(logits, yb)
        running_loss += loss.item() * xb.size(0)
        n += xb.size(0)
        all_logits.append(logits.detach().cpu())
        all_targets.append(yb.detach().cpu())
    loss = running_loss / max(n, 1)
    logits = torch.cat(all_logits)
    targets = torch.cat(all_targets)
    probs = torch.sigmoid(logits).numpy()
    # print(targets.numpy())
    y_true = targets.numpy().astype(int)
    # print(y_true)
    return loss, probs, y_true


def train_eval_mlp_torch(
    X_train,
    y_train,
    X_test,
    y_test,
    hidden=(128, 64),
    lr=1e-3,
    weight_decay=1e-4,
    batch_size=256,
    max_epochs=100,
    patience=10,
    val_size=0.1,
    seed=42,
    verbose=False,
):
    """
    Train a 2-layer MLP (no scaling) with early stopping on a val split from X_train.
    Returns: (test_auc, y_prob_test)
    """
    set_seed(seed)

    # Make a stratified val split from the given training data
    X_tr, X_val, y_tr, y_val = train_test_split(
        X_train, y_train, test_size=val_size, stratify=y_train, random_state=seed
    )

    # Ensure dtypes
    X_tr = X_tr.astype(np.float32)
    X_val = X_val.astype(np.float32)
    X_te = X_test.astype(np.float32)
    y_trf = y_tr.astype(np.float32)
    y_valf = y_val.astype(np.float32)
    y_te = y_test.astype(int)

    # Datasets / loaders
    train_ds = TensorDataset(torch.from_numpy(X_tr), torch.from_numpy(y_trf))
    val_ds = TensorDataset(torch.from_numpy(X_val), torch.from_numpy(y_valf))
    test_ds = TensorDataset(
        torch.from_numpy(X_te), torch.from_numpy(y_te.astype(np.float32))
    )

    train_loader = DataLoader(
        train_ds, batch_size=batch_size, shuffle=True, pin_memory=False
    )
    val_loader = DataLoader(
        val_ds, batch_size=batch_size, shuffle=False, pin_memory=False
    )
    test_loader = DataLoader(
        test_ds, batch_size=batch_size, shuffle=False, pin_memory=False
    )

    # Model / loss / optim
    in_dim = X_train.shape[1]
    h1, h2 = hidden
    model = MLP(in_dim=in_dim, h1=h1, h2=h2).to(DEVICE)
    criterion = nn.BCEWithLogitsLoss()
    optim = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)

    # Early stopping
    best_state = None
    best_val_loss = float("inf")
    epochs_no_improve = 0

    for epoch in range(1, max_epochs + 1):
        model.train()
        running = 0.0
        n_seen = 0
        for xb, yb in train_loader:
            xb = xb.to(DEVICE, non_blocking=True)
            yb = yb.to(DEVICE, non_blocking=True)
            optim.zero_grad()
            logits = model(xb)
            loss = criterion(logits, yb)
            loss.backward()
            optim.step()
            running += loss.item() * xb.size(0)
            n_seen += xb.size(0)
        train_loss = running / max(n_seen, 1)

        val_loss, _, _ = _evaluate(model, val_loader, criterion)

        if verbose:
            print(
                f"Epoch {epoch:03d}  train_loss={train_loss:.4f}  val_loss={val_loss:.4f}"
            )

        if val_loss + 1e-5 < best_val_loss:
            best_val_loss = val_loss
            best_state = {
                k: v.detach().cpu().clone() for k, v in model.state_dict().items()
            }
            epochs_no_improve = 0
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                if verbose:
                    print(
                        f"Early stopping at epoch {epoch} (best val_loss={best_val_loss:.4f})"
                    )
                break

    if best_state is not None:
        model.load_state_dict({k: v.to(DEVICE) for k, v in best_state.items()})

    # Final test evaluation
    _, y_prob_test, y_true_test = _evaluate(model, test_loader, criterion)
    # print(y_true_test.min())
    # print(y_prob_test.min())
    # print(y_prob_test.max())
    # print(y_true_test.shape)
    # print(y_prob_test.shape)
    test_auc = roc_auc_score(y_true_test, y_prob_test)

    return test_auc, y_prob_test
```

```python
classification_aucs_200 = np.zeros((8, 8))

with open('project_path.txt') as f:
    local_project_path = f.read()
    
for i, j in tqdm(combinations(range(8), 2), total=len(list(combinations(range(8), 2)))):
    self_matrix_1_file = local_project_path + f"medium_w2/w2_block_{i}_{i}.npy"
    self_matrix_2_file = local_project_path + f"medium_w2/w2_block_{j}_{j}.npy"
    off_diag_matrix_file = local_project_path + f"medium_w2/w2_block_{i}_{j}.npy"

    big_matrix = ot_utils.combine_distance_blocks(
        [self_matrix_1_file, off_diag_matrix_file, self_matrix_2_file]
    )

    labels = np.repeat([0, 1], 4000)

    rel_reps_w2 = comparison_utils.relative_rep(
        big_matrix, labels, num_classes=2, anchors_per_class=100, seed=42
    )
    
    rel_reps_w2 *= 10
    
    self_matrix_1_file = local_project_path + f"medium_ps/ps_block_{i}_{i}.npy"
    self_matrix_2_file = local_project_path + f"medium_ps/ps_block_{j}_{j}.npy"
    off_diag_matrix_file = local_project_path + f"medium_ps/ps_block_{i}_{j}.npy"

    big_matrix = ot_utils.combine_distance_blocks(
        [self_matrix_1_file, off_diag_matrix_file, self_matrix_2_file]
    )

    labels = np.repeat([0, 1], 4000)

    rel_reps_ps = comparison_utils.relative_rep(
        big_matrix, labels, num_classes=2, anchors_per_class=100, seed=42
    )
    
    rel_reps = np.concatenate((rel_reps_w2, rel_reps_ps), axis=1)
    # print(rel_reps.shape)

    y = np.r_[np.zeros(4000, dtype=int), np.ones(4000, dtype=int)]

    # Train/test split (stratified)
    X_train, X_test, y_train, y_test = train_test_split(
        rel_reps, y, test_size=0.2, stratify=y, random_state=42
    )

    # --- PyTorch MLP instead of sklearn GridSearch ---
    auc, _ = train_eval_mlp_torch(
        X_train,
        y_train,
        X_test,
        y_test,
        hidden=(128, 64),
        lr=1e-4,
        weight_decay=0,
        batch_size=256,
        max_epochs=100,
        patience=10,
        val_size=0.1,
        seed=42,
        verbose=False,  # set True if you want per-epoch logs
    )

    classification_aucs_200[i, j] = auc
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
from matplotlib.cm import ScalarMappable

# ---------- Helpers ----------
def upper_tri_to_full(rows, n=8):
    """Fill an n×n matrix (NaN on/under diag) from upper-tri rows list."""
    M = np.full((n, n), np.nan, dtype=float)
    for i, row_vals in enumerate(rows):           # i=0..n-2 -> row 0..n-2
        r = i
        for j, v in enumerate(row_vals, start=r+1):
            M[r, j] = v
    return M

def annotate_upper_tri(ax, M):
    n = M.shape[0]
    for i in range(n):
        for j in range(n):
            if not np.isnan(M[i, j]):
                ax.text(j, i, f"{M[i, j]:.3f}", ha="center", va="center", fontsize=12, color = 'w')

def heat(ax, M, title, norm):
    im = ax.imshow(M, norm=norm)
    ax.set_title(title, pad=8)
    ax.set_xlabel("Class")
    ax.set_ylabel("Class")
    ax.set_xticks(range(8)); ax.set_yticks(range(8))
    ax.set_xticklabels(range(1, 9)); ax.set_yticklabels(range(1, 9))
    annotate_upper_tri(ax, M)
    return im

# ---------- Manually extracted values ----------
# Joint Relative Space
joint_rows = [
    [0.987, 0.990, 0.984, 0.984, 0.969, 0.961, 0.999],
    [0.534, 0.773, 0.881, 0.799, 0.869, 0.931],
    [0.822, 0.914, 0.807, 0.874, 0.930],
    [0.668, 0.889, 0.919, 0.890],
    [0.917, 0.941, 0.910],
    [0.622, 0.976],
    [0.973],
]
joint = upper_tri_to_full(joint_rows, 8)

# Relative Space w2
w2_rows = [
    [0.985, 0.987, 0.985, 0.984, 0.966, 0.963, 0.998],
    [0.539, 0.781, 0.874, 0.788, 0.859, 0.921],
    [0.829, 0.910, 0.805, 0.873, 0.921],
    [0.690, 0.890, 0.925, 0.887],
    [0.916, 0.942, 0.902],
    [0.624, 0.976],
    [0.971],
]
w2 = upper_tri_to_full(w2_rows, 8)

# Relative Space ps
ps_rows = [
    [0.979, 0.982, 0.983, 0.983, 0.940, 0.925, 0.999],
    [0.517, 0.512, 0.623, 0.797, 0.867, 0.897],
    [0.529, 0.578, 0.805, 0.874, 0.882],
    [0.508, 0.820, 0.886, 0.890],
    [0.805, 0.875, 0.891],
    [0.622, 0.971],
    [0.971],
]
ps = upper_tri_to_full(ps_rows, 8)

# ---------- Uniform color scale across all panels ----------
vals_all = np.concatenate([joint[~np.isnan(joint)],
                           w2[~np.isnan(w2)],
                           ps[~np.isnan(ps)]])
norm = Normalize(vmin=float(vals_all.min()), vmax=float(vals_all.max()))

# ---------- Plot: side-by-side with shared colorbar ----------
fig, axes = plt.subplots(1, 3, figsize=(18, 5), constrained_layout=True)
im1 = heat(axes[0], joint, "Joint Relative Space", norm)
im2 = heat(axes[1], w2,   "Relative Space w2",    norm)
im3 = heat(axes[2], ps,   "Relative Space ps",    norm)

# One shared colorbar
cbar = fig.colorbar(ScalarMappable(norm=norm), ax=axes.ravel().tolist(), shrink=0.9)
cbar.set_label("AUC")

# plt.savefig("pairwise_auc_side_by_side.png", dpi=200, bbox_inches="tight")
plt.show()
```
