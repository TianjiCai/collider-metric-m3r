```python
import numpy as np
from tqdm import tqdm
import ot_utils

file = np.load("all_jets_10_subjets_p4s.npz")

all_jets_transformed_p4s = file["X_transformed_epxpypz"]
```

```python
all_jets_transformed_p4s.shape
```

```python
ps_dist_01 = ot_utils.PSdistance(
    all_jets_transformed_p4s[0],
    all_jets_transformed_p4s[1],
    Sphere2Simplex=1,
    l_has_chi=True,
)
ps_dist_01
```

```python
ps_dist_10 = ot_utils.PSdistance(
    all_jets_transformed_p4s[1],
    all_jets_transformed_p4s[0],
    Sphere2Simplex=1,
    l_has_chi=True,
)
ps_dist_10
```

```python
n_samples = 8000
ps_dist_matrix = np.zeros((n_samples, n_samples))

i_pairs, j_pairs = np.triu_indices(n_samples, k=1)

for i, j in tqdm(zip(i_pairs, j_pairs), total=len(i_pairs)):
    ps_dist_matrix[i, j] = ot_utils.PSdistance(
        all_jets_transformed_p4s[i],
        all_jets_transformed_p4s[j],
        Sphere2Simplex=1,
        l_has_chi=True,
        return_only_l=True,
    )

    ps_dist_matrix[j, i] = ps_dist_matrix[i, j]

ps_dist_matrix_2 = ot_utils.parallel_dist_mat_square(
    event_set=all_jets_transformed_p4s,
    dist_func=ot_utils.PSdistance,
    events_per_process=8192,
    Sphere2Simplex=1,
    l_has_chi=True,
    return_only_l=True,
)

assert np.allclose(ps_dist_matrix, ps_dist_matrix_2)
```
